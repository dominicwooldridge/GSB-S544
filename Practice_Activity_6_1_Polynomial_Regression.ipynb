```python
from sklearn.preprocessing import OneHotEncoder
import pandas as pd
import numpy as np
from palmerpenguins import penguins
from palmerpenguins import load_penguins
penguins = load_penguins()
```

```python
penguins.dropna(inplace=True)
pd.get_dummies(penguins)
```

Let's use `bill_length_mm` to predict `bill_depth_mm`. Prepare your data and fit the following models on the entire dataset:

* Simple linear regression (e.g. straight-line) model
* Quadratic (degree 2 polynomial) model
* Cubic (degree 3 polynomial) model
* Degree 10 polynomial model

Make predictions for each model and plot your fitted models on the scatterplot.

```python
# Code Here

penguins['bill_length_sq'] = penguins["bill_length_mm"]**2
penguins['bill_length_cub'] = penguins["bill_length_mm"]**3
penguins['bill_length_4'] = penguins["bill_length_mm"]**4
penguins['bill_length_5'] = penguins["bill_length_mm"]**5
penguins['bill_length_6'] = penguins["bill_length_mm"]**6
penguins['bill_length_7'] = penguins["bill_length_mm"]**7
penguins['bill_length_8'] = penguins["bill_length_mm"]**8
penguins['bill_length_9'] = penguins["bill_length_mm"]**9
penguins['bill_length_10'] = penguins["bill_length_mm"]**10
```

```python
from sklearn.linear_model import LinearRegression
X = penguins[["bill_length_mm"]]
Y = penguins["bill_depth_mm"]
model = LinearRegression()
model.fit(X,Y)
model.predict(X)
```

```python
X = penguins[["bill_length_mm", "bill_length_sq"]]
Y = penguins["bill_depth_mm"]
model_2 = LinearRegression()
model_2.fit(X,Y)
model_2.predict(X)
```

```python
X = penguins[["bill_length_mm", "bill_length_sq", "bill_length_cub"]]
Y = penguins["bill_depth_mm"]
model_3 = LinearRegression()
model_3.fit(X,Y)
model_3.predict(X)
```

```python
X = penguins[["bill_length_mm", "bill_length_sq", "bill_length_cub", "bill_length_4", "bill_length_5", "bill_length_6", "bill_length_7", "bill_length_8", "bill_length_9", "bill_length_10"]]
Y = penguins["bill_depth_mm"]
model_10 = LinearRegression()
model_10.fit(X,Y)
model_10.predict(X)
```

```python
import numpy as np
import pandas as pd

def X_for_degree(df, d):
    cols = []
    for k in range(1, d+1):
        if k == 1:
            cols.append('bill_length_mm')
        elif k == 2:
            cols.append('bill_length_sq')
        elif k == 3:
            cols.append('bill_length_cub')
        else:
            cols.append(f'bill_length_{k}')
    return df[cols]

model_cols = {
    1: 'model',
    2: 'model_2',
    3: 'model_3',
    10: 'model_10',
}


for degree, mcol in model_cols.items():
    if mcol not in penguins.columns:
        continue


    m = penguins[mcol].dropna().iloc[0]

    X_d = X_for_degree(penguins, degree)
    mask = X_d.notna().all(axis=1)

    pred_col = f'pred_deg{degree}'
    penguins[pred_col] = np.nan
    penguins.loc[mask, pred_col] = m.predict(X_d[mask])

penguins = penguins.drop(columns=list(model_cols.values()), errors='ignore')
```

```python
from plotnine import ggplot, aes, geom_point, geom_line, theme_minimal, labs

(
    ggplot(penguins, aes('bill_length_mm', 'bill_depth_mm'))
    + geom_point(alpha=0.4)
    + geom_line(aes(y='pred_deg1'), color='Red')
    + geom_line(aes(y='pred_deg2'), color='Green')
    + geom_line(aes(y='pred_deg3'), color='Blue')
    + geom_line(aes(y='pred_deg10'), color='Orange')
    + labs(x='Bill length (mm)', y='Bill depth (mm)', title='Polynomial fits')
    + theme_minimal()
)
```
