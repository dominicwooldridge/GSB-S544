1. Write a function that takes in information about a penguin, and returns one of the following definitions:

*   "Big Mouth Billy": Male penguins with bill length times bill depth greater than 800.
*   "Dainty Daisy": Female penguins with flipper length less than 5% of body mass.
*   "Average Adelie:" Any Adelie penguins that do not fall into either category.
*   "Other": Any penguins that do not fall into any of the categories.



```python
penguins = palmerpenguins.load_penguins()
penguins.head()
```

```python
def big_mouth_billy(penguins):
  condition = (penguins["sex"] == "male") & (penguins["bill_length_mm"] * penguins["bill_depth_mm"] > 800)
  return penguins[condition]
```

```python
def dainty_daisy(penguins):
  condition = (penguins["sex"] == "female") & (penguins["flipper_length_mm"] < 0.05 * penguins["body_mass_g"])
  return penguins[condition]
```

```python
def average_adelie(penguins):
  condition_billy = (penguins["sex"] == "male") & (penguins["bill_length_mm"] * penguins["bill_depth_mm"] > 800)
  condition_daisy = (penguins["sex"] == "female") & (penguins["flipper_length_mm"] < 0.05 * penguins["body_mass_g"])
  condition_adelie = (penguins["species"] == "Adelie") & ~(condition_billy) & ~(condition_daisy)
  return penguins[condition_adelie]
```

```python
def other(penguins):
  condition_billy = (penguins["sex"] == "male") & (penguins["bill_length_mm"] * penguins["bill_depth_mm"] > 800)
  condition_daisy = (penguins["sex"] == "female") & (penguins["flipper_length_mm"] < 0.05 * penguins["body_mass_g"])
  condition_adelie = (penguins["species"] == "Adelie") & ~(condition_billy) & ~(condition_daisy)
  condition_other = ~(condition_billy) & ~(condition_daisy) & ~(condition_adelie)
  return penguins[condition_other]
```

2. Use an iterable function to create a new variable called `category_name` that adds these labels.

```python
def category_name(row):
  if (row["sex"] == "male") & (row["bill_length_mm"] * row["bill_depth_mm"] > 800):
    return "Big Mouth Billy"

  elif (row["sex"] == "female") & (row["flipper_length_mm"] < 0.05 * row["body_mass_g"]):
    return "Dainty Daisy"

  elif (row["species"] == "Adelie"):
    return "Average Adelie"

  else:
    return "Other"

penguins["category_name"] = penguins.apply(category_name, axis=1)
```
